```python
import pandas as pd
import matplotlib.pyplot as plt
from helpers import load_data, clean_data, print_quality_report

raw = load_data()
df, report = clean_data(raw)
print_quality_report(report)
```

```python
# Total revenue
total_revenue = df["revenue"].sum()
print(f"Total revenue: R{total_revenue:,.2f}")

# Revenue per category, ranked highest to lowest
revenue_by_category = (
    df.groupby("category")["revenue"].sum().sort_values(ascending=False)
)
print("\nRevenue by category:")
print(revenue_by_category)

# Revenue per store
revenue_by_store = (
    df.groupby("store")["revenue"].sum().sort_values(ascending=False)
)
print("\nRevenue by store:")
print(revenue_by_store)
```

```python
# Best-selling product by quantity vs by revenue
qty_by_product = (
    df.groupby("product_name")["quantity"].sum().sort_values(ascending=False)
)
revenue_by_product = (
    df.groupby("product_name")["revenue"].sum().sort_values(ascending=False)
)
print("Top product by quantity:")
print(qty_by_product.head(3))
print("\nTop product by revenue:")
print(revenue_by_product.head(3))

# Revenue month by month
df["month"] = df["date"].dt.to_period("M")
revenue_by_month = df.groupby("month")["revenue"].sum()
print("\nRevenue by month:")
print(revenue_by_month)

# Average transaction value
avg_transaction = df["revenue"].mean()
print(f"\nAverage transaction value: R{avg_transaction:,.2f}")

# Most common payment method
payment_counts = df["payment_method"].value_counts()
print("\nPayment method counts:")
print(payment_counts)
```

```python
import matplotlib.pyplot as plt

# Chart 1: Revenue by category (bar)
plt.figure(figsize=(8, 5))
revenue_by_category.plot(kind="bar", color="steelblue")
plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue (R)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("charts/revenue_by_category.png")
plt.show()
```

```python
# Chart 2: Revenue by store (bar)
plt.figure(figsize=(8, 5))
revenue_by_store.plot(kind="bar", color="darkorange")
plt.title("Revenue by Store")
plt.xlabel("Store")
plt.ylabel("Revenue (R)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("charts/revenue_by_store.png")
plt.show()
```

```python
# Chart 3: Revenue by month (line)
plt.figure(figsize=(8, 5))
revenue_by_month.plot(kind="line", marker="o", color="seagreen")
plt.title("Revenue by Month")
plt.xlabel("Month")
plt.ylabel("Revenue (R)")
plt.tight_layout()
plt.savefig("charts/revenue_by_month.png")
plt.show()
```
